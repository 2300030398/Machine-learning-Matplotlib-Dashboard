```python
import pandas as pd
import missingno as ms
online_shopping=pd.DataFrame({
    "Customer_name":["ashi","teja","harshini"],
    "purchesd":[10,10,1],
    "returned":[10,2,None]
    
})
k=online_shopping.dropna(how='all')
print(k)
print()
k=online_shopping.dropna()#if any null value means 
print(k)
print()
k=online_shopping.dropna(subset=['Customer_name','returned'])#if any two of the cols have null val then we will leave it
print(k)
print()
k=online_shopping.dropna(thresh=3)#Keep rows having at least 3 non-missing values  if 
print(k)
```

```python
k=online_shopping.fillna(0)

print(k)
print()
k=online_shopping.fillna(1)
print(k)
print()
k=online_shopping.ffill()
print(k)
print()
k=online_shopping.bfill()
print(k)
```

```python
online_shopping['returned']=online_shopping['returned'].fillna(online_shopping['returned'].mean())
print(online_shopping)
```

```python
online_shopping['returned']=online_shopping['returned'].fillna(online_shopping['returned'].mode())
print(online_shopping)
```

```python
online_shopping['returned']=online_shopping['returned'].fillna(online_shopping['returned'].median())
print(online_shopping)
```

```python
online_shopping.drop_duplicates()
print(online_shopping)
online_shopping.drop_duplicates(keep='first')
online_shopping.drop_duplicates(keep='last')
online_shopping.drop_duplicates(keep=False)
d=online_shopping.duplicated('purchased').sum()
print(d)
```
