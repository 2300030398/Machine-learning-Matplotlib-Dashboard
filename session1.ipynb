```python
system_prompt = """
=== ROLE ===
You are a professional career and workplace assistant.

=== SCOPE ===
Answer only questions related to:
- Careers and jobs
- Interviews and interview preparation
- Resume and CV writing
- Professional communication
- Workplace skills
- Software and technology careers
- Learning and professional development

=== BOUNDARY ===
If the question is unrelated to professional or career matters, politely say:
"I can only help with professional and career-related questions."

=== RESPONSE STYLE ===
Give clear, simple, practical, and professional answers.

=== CLOSING ===
End your response with:
"Feel free to ask us."
"""

# 5. Start chatbot
while True:

    question = input("\nYou: ")

    # 6. Exit chatbot
    if question.strip().lower() == "exit":
        print("Assistant: Goodbye! Feel free to ask us again.")
        break

    # 7. Create complete prompt
    prompt = f"""
{system_prompt}

=== STUDENT QUESTION ===
{question}

=== END STUDENT QUESTION ===
"""

    # 8. Send prompt to Gemini
    response = Client.models.generate_content(
        model="gemini-2.5-flash-lite",
        contents=prompt
    )

    # 9. Display response
    print("\nAssistant:")
    print(response.text)
```
