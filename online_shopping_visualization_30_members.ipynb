# Online Shopping Data Visualization Practice

**Example dataset:** 30 online shoppers. This notebook practices pandas, Matplotlib and Seaborn charts, from simple plots to dashboards.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
# Create a sample dataset for 30 online shoppers
data = {
    "Member": [f"Member {i}" for i in range(1, 31)],
    "Age": [21, 25, 32, 28, 22, 35, 29, 24, 31, 27, 23, 38, 26, 34, 30,
            22, 29, 33, 25, 36, 28, 24, 31, 27, 39, 26, 32, 23, 35, 29],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male"],
    "Category": ["Fashion", "Electronics", "Beauty", "Fashion", "Grocery", "Electronics", "Books", "Fashion",
                 "Beauty", "Grocery", "Electronics", "Fashion", "Books", "Beauty", "Electronics", "Grocery",
                 "Fashion", "Books", "Beauty", "Electronics", "Grocery", "Fashion", "Electronics", "Beauty",
                 "Books", "Grocery", "Fashion", "Electronics", "Beauty", "Books"],
    "Orders": [3, 5, 2, 4, 6, 2, 3, 5, 4, 2, 7, 3, 1, 4, 6, 3, 5, 2, 4, 3, 6, 2, 5, 4, 1, 3, 6, 2, 4, 5],
    "Amount_Spent": [2400, 18500, 1200, 3600, 2800, 22000, 900, 4500, 1600, 3100,
                     19500, 5200, 700, 1400, 26000, 3300, 6100, 1100, 1800, 17500,
                     2900, 4100, 23000, 1500, 800, 2700, 5800, 21000, 1700, 1300],
    "Visits_Per_Month": [8, 12, 6, 10, 9, 7, 5, 11, 8, 6, 14, 9, 4, 7, 13, 8, 12, 5, 7, 10, 9, 6, 12, 8, 3, 7, 11, 9, 6, 10],
    "Satisfaction": [4, 5, 3, 4, 4, 5, 3, 4, 4, 3, 5, 4, 3, 4, 5, 4, 5, 3, 4, 5, 4, 3, 5, 4, 3, 4, 5, 4, 3, 4]
}

df = pd.DataFrame(data)
df.to_csv("online_shopping_data.csv", index=False)
df.head()
```

```python
# View all 30 members and basic information
print("Dataset shape:", df.shape)
display(df)
df.info()
```

## 1. Bar chart from the CSV

Compare the amount spent by each online shopping member.

```python
plt.figure(figsize=(13, 5))
plt.bar(df["Member"], df["Amount_Spent"])
plt.title("Amount Spent by Each Online Shopper")
plt.xlabel("Member")
plt.ylabel("Amount Spent (₹)")
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

## 2. Basic Matplotlib charts

These examples use small lists typed directly into the notebook.

```python
# Line plot: monthly orders for a sample shopper
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
orders = [2, 3, 4, 3, 5, 6]

plt.plot(months, orders, marker="o")
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Bar chart: spending by four sample shoppers
members = ["Asha", "Ravi", "Maya", "Arun"]
spending = [2500, 5200, 3400, 6100]

plt.bar(members, spending)
plt.title("Spending by Four Shoppers")
plt.xlabel("Shopper")
plt.ylabel("Amount Spent (₹)")
plt.show()
```

```python
# Histogram: order amounts for 20 sample transactions
transaction_amounts = [450, 700, 850, 900, 1100, 1200, 1300, 1500, 1700, 1800,
                       2100, 2300, 2500, 2700, 3000, 3400, 3900, 4500, 5200, 6100]

plt.hist(transaction_amounts, bins=6, edgecolor="black")
plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Frequency")
plt.show()
```

## 3. One dataset shown in five chart types

The following five charts use the same 10 days of average online order values.

```python
days = list(range(1, 11))
average_order_value = [850, 920, 780, 1100, 1250, 980, 1350, 1180, 1420, 1550]

# Line plot
plt.plot(days, average_order_value, marker="*", markersize=10)
plt.title("Average Order Value Across 10 Days")
plt.xlabel("Day")
plt.ylabel("Average Order Value (₹)")
plt.show()

# Bar chart
plt.bar(days, average_order_value)
plt.title("Average Order Value - Bar Chart")
plt.xlabel("Day")
plt.ylabel("Average Order Value (₹)")
plt.show()

# Histogram
plt.hist(average_order_value, bins=5, edgecolor="black")
plt.title("Distribution of Average Order Values")
plt.xlabel("Average Order Value (₹)")
plt.ylabel("Frequency")
plt.show()

# Pie chart
plt.pie(average_order_value, labels=[f"Day {d}" for d in days], autopct="%1.1f%%")
plt.title("Share of Total Value by Day")
plt.show()

# Scatter plot
plt.scatter(days, average_order_value)
plt.title("Day vs Average Order Value")
plt.xlabel("Day")
plt.ylabel("Average Order Value (₹)")
plt.show()
```

## 4. Subplots and multiple plots in a single figure

### Student-style dashboard adapted to online shopping

This 3×2 dashboard summarizes category spending, gender split, visits versus spending, spending distribution, shoppers per category and visit distribution.

```python
fig, axes = plt.subplots(3, 2, figsize=(14, 15))
fig.suptitle("Online Shopping Performance Dashboard", fontsize=16)

category_spend = df.groupby("Category")["Amount_Spent"].mean().sort_values(ascending=False)
axes[0, 0].bar(category_spend.index, category_spend.values)
axes[0, 0].set_title("Average Spending by Category")
axes[0, 0].set_ylabel("Average Spend (₹)")
axes[0, 0].tick_params(axis="x", rotation=25)

gender_counts = df["Gender"].value_counts()
axes[0, 1].pie(gender_counts.values, labels=gender_counts.index, autopct="%1.1f%%")
axes[0, 1].set_title("Gender Split")

axes[1, 0].scatter(df["Visits_Per_Month"], df["Amount_Spent"])
axes[1, 0].set_title("Monthly Visits vs Amount Spent")
axes[1, 0].set_xlabel("Visits per Month")
axes[1, 0].set_ylabel("Amount Spent (₹)")

axes[1, 1].hist(df["Amount_Spent"], bins=8, edgecolor="black")
axes[1, 1].set_title("Spending Distribution")
axes[1, 1].set_xlabel("Amount Spent (₹)")

category_counts = df["Category"].value_counts()
axes[2, 0].bar(category_counts.index, category_counts.values)
axes[2, 0].set_title("Number of Shoppers by Category")
axes[2, 0].set_ylabel("Members")
axes[2, 0].tick_params(axis="x", rotation=25)

axes[2, 1].hist(df["Visits_Per_Month"], bins=7, edgecolor="black")
axes[2, 1].set_title("Monthly Website Visit Distribution")
axes[2, 1].set_xlabel("Visits per Month")

plt.tight_layout()
plt.show()
```

### Seaborn scatter plot

Explore how monthly visits and spending relate, using product category as the colour and gender as the marker style.

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="Visits_Per_Month", y="Amount_Spent",
                hue="Category", style="Gender", s=100)
plt.title("Website Visits vs Spending by Category and Gender")
plt.xlabel("Visits per Month")
plt.ylabel("Amount Spent (₹)")
plt.show()
```

### 2×2 Online Shopping Analysis dashboard

The charts show spending distribution, shoppers by category, spending by gender and average spending by category.

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle("Online Shopping Analysis", fontsize=16)

sns.histplot(data=df, x="Amount_Spent", bins=8, ax=axes[0, 0])
axes[0, 0].set_title("Spending Distribution")

sns.countplot(data=df, x="Category", ax=axes[0, 1])
axes[0, 1].set_title("Shoppers per Category")
axes[0, 1].tick_params(axis="x", rotation=25)

sns.boxplot(data=df, x="Gender", y="Amount_Spent", ax=axes[1, 0])
axes[1, 0].set_title("Spending by Gender")

sns.barplot(data=df, x="Category", y="Amount_Spent", errorbar=None, ax=axes[1, 1])
axes[1, 1].set_title("Average Spending by Category")
axes[1, 1].tick_params(axis="x", rotation=25)

plt.tight_layout()
plt.show()
```

### Final 2×2 dashboard: shopping trends

A line chart, category pie chart, bar chart and scatter plot using the sample shopping data.

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle("Online Shopping Trends Dashboard", fontsize=16)

# Line chart: average spend by age
age_spend = df.groupby("Age")["Amount_Spent"].mean().sort_index()
axes[0, 0].plot(age_spend.index, age_spend.values, marker="o")
axes[0, 0].set_title("Average Spending by Age")
axes[0, 0].set_xlabel("Age")
axes[0, 0].set_ylabel("Average Spend (₹)")

# Pie chart: category share by number of members
category_counts = df["Category"].value_counts()
axes[0, 1].pie(category_counts.values, labels=category_counts.index, autopct="%1.1f%%")
axes[0, 1].set_title("Members by Shopping Category")

# Bar chart: total orders by category
orders_by_category = df.groupby("Category")["Orders"].sum().sort_values(ascending=False)
axes[1, 0].bar(orders_by_category.index, orders_by_category.values)
axes[1, 0].set_title("Total Orders by Category")
axes[1, 0].set_ylabel("Orders")
axes[1, 0].tick_params(axis="x", rotation=25)

# Scatter plot: satisfaction vs amount spent
axes[1, 1].scatter(df["Satisfaction"], df["Amount_Spent"])
axes[1, 1].set_title("Satisfaction vs Amount Spent")
axes[1, 1].set_xlabel("Satisfaction (1-5)")
axes[1, 1].set_ylabel("Amount Spent (₹)")

plt.tight_layout()
plt.show()
```
